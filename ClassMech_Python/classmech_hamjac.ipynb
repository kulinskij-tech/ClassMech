# The Hamilton–Jacobi equation and separation of variables

**Draft lecture.** In an oscillator, energy fixes the amplitude; in a central field, angular momentum reduces the orbit problem to radial motion. The Hamilton–Jacobi method organizes such reductions through a single scalar action. We shall use additive separation to turn a first-order PDE into one-dimensional quadratures, and then recover trajectories by differentiating with respect to the separation constants.

Prerequisites: Hamilton's equations, canonical transformations, and [characteristics of first-order PDEs](https://colab.research.google.com/github/kulinskij-tech/ClassMech/blob/main/ClassMech_Python/classmech_pdechar.ipynb). The worked examples below develop the method before the retained canonical-transformation background at the end.

## The action and the general additive scheme

For configuration coordinates $q^i$ and their conjugate momenta $p_i$, Hamilton's principal function satisfies

$$
\frac{\partial S}{\partial t}+H\left(q^i,\frac{\partial S}{\partial q^i},t\right)=0,\qquad p_i=\frac{\partial S}{\partial q^i}.
$$

For an autonomous Hamiltonian, look for $S=W-Et$, i.e. time dependence splits. This give a hint to the general method of splitting any cyclic variable and integration of mechanical problems via Hamilton-Javoby equation.   
Full additive separation means finding coordinates in which
<!--  -->
$$
W(q;\alpha)=\sum_{i=1}^{n}W_i(q^i;\alpha_1,\ldots,\alpha_n),\qquad H\left(q,\partial_qW\right)=E.
$$
<!--  -->
The constants are shared by the separated terms; the terms need not each depend on just one constant. Choose $\alpha_1=E$ locally when this is a regular parametrization. A **complete integral** contains $n$ independent essential parameters, with
<!--  -->
$$
\det\left(\frac{\partial^2W}{\partial q^i\partial\alpha_a}\right)\ne0
$$
<!--  -->
on the coordinate patch being used. An arbitrary additive constant in $S$ is irrelevant and is not one of these $n$ essential parameters. A single solution at one fixed energy is generally not a complete integral.

The practical scheme is:

1. Identify cyclic coordinates and conserved momenta; choose adapted coordinates.
2. Substitute an additive ansatz into the stationary HJ equation.
3. Separate coordinate-dependent terms using constants and solve for each momentum branch.
4. Integrate $W_i=\int p_i\,dq^i$; elementary antiderivatives are not required.
5. Recover the motion from $p_i=\partial_iW$ and $\beta_a=\partial S/\partial\alpha_a$.

In particular,

$$
\frac{\partial W}{\partial E}=t+\beta_E,\qquad \frac{\partial W}{\partial\alpha_a}=\beta_a\quad(a>1).
$$

Differentiation uses consistent local branches and reference points. Turning points require patching the two momentum signs; a single generating function is usually local. Additive separation depends on the coordinates and is not guaranteed for an arbitrary Hamiltonian. A separation constant must genuinely be independent of the coordinate being separated; merely renaming a coordinate-dependent expression does not achieve separation.

## Why separation solves the characteristic problem

The Hamilton–Jacobi equation is a first-order PDE. The trajectories recovered by differentiating a complete integral are exactly its characteristic curves. To see this, apply the general characteristic scheme from the [first-order PDE lecture](https://colab.research.google.com/github/kulinskij-tech/ClassMech/blob/main/ClassMech_Python/classmech_pdechar.ipynb), with an independent curve parameter s:

$$
F(t,q,S,e,p)=e+H(q,p,t)=0,\qquad e=S_t,\qquad p_i=S_{q^i}.
$$

Here e is the gradient component conjugate to time, not the mechanical energy. The Charpit equations give

$$
\frac{dt}{ds}=F_e=1,\qquad \frac{dq^i}{ds}=F_{p_i}=\frac{\partial H}{\partial p_i},\qquad
\frac{dp_i}{ds}=-F_{q^i}-p_iF_S=-\frac{\partial H}{\partial q^i},
$$

$$
\frac{de}{ds}=-\frac{\partial H}{\partial t},\qquad
\frac{dS}{ds}=eF_e+p_iF_{p_i}=p_i\frac{\partial H}{\partial p_i}-H.
$$

Because t grows at unit rate, it can be used as the characteristic parameter. The middle two equations are Hamilton's equations, and the last one says **dS/dt=L** along a mechanical trajectory. The principal function itself is generally not constant along a characteristic. For an autonomous Hamiltonian, e=-E is constant, which motivates S=W-Et.

### Direct check using the gradient of S

On a local smooth solution surface, set $p_i=S_{q^i}$. Differentiate the HJ PDE with respect to $q^i$ and follow a curve with $\dot q^i=H_{p_i}$:

$$
\frac{dp_i}{dt}=S_{tq^i}+S_{q^iq^j}\dot q^j
=S_{tq^i}+S_{q^iq^j}H_{p_j}=-H_{q^i}.
$$

Thus the gradient of the action evolves as the mechanical momentum. This is also why a smooth solution graph is preserved locally by the characteristic flow. It need not remain a single-valued graph after a caustic or a turning-point projection singularity; different branches of the action may then be necessary.

### Why the parameter derivatives label characteristics

Let $S(q,t;\alpha)$ be a complete integral whose parameters do not occur explicitly in the Hamiltonian. Differentiating the HJ PDE with respect to an essential parameter gives

$$
\partial_t S_{\alpha_a}+H_{p_i}\partial_{q^i}S_{\alpha_a}=0.
$$

This is a transport equation along the mechanical characteristic field. Consequently,

$$
\frac{d}{dt}\left(S_{\alpha_a}\right)=0,\qquad S_{\alpha_a}=\beta_a.
$$

The $\beta_a$ are constants along a characteristic, while the $\alpha_a$ label the complete-integral family. Together, on a regular patch, they provide the integration constants for the mechanical motion. In the autonomous separated solution, $S_E=W_E-t=\beta_E$ is constant: **W_E itself increases like time**, rather than being constant.

### The same construction in the four examples

- **Oscillator:** on a positive-momentum branch, $W_E=\arcsin(q/A)/\omega$. Setting $W_E=t+\beta_E$ parametrizes precisely the Hamiltonian ellipse; changing the momentum branch continues the same characteristic through a turning point.
- **Pendulum:** $\partial_EW=\int I\,d\theta/p_\theta$ is the characteristic relation $dt=I\,d\theta/p_\theta$, since $\dot\theta=p_\theta/I$.
- **Central field:** $W_E$ recovers radial time, and $S_\ell=\varphi+W_{r,\ell}$ is constant. Eliminating time yields the same relation $d\varphi/dr=\ell/(r^2p_r)$ as the ratio of Hamilton's equations.
- **Kepler:** integrating those radial and angular characteristic relations gives the conic and Kepler's time equation. They describe one characteristic in two different parameterizations, not two different solutions.

Additive separation therefore supplies first integrals and quadratures for the characteristic ODEs. Direct numerical integration of Hamilton's equations, as used in the comparisons below, traces those same characteristic curves without first constructing the complete integral.

## 2. Harmonic oscillator: the simplest complete integral

Take $m>0$, $\omega>0$ and

$$
H=\frac{p^2}{2m}+\frac{m\omega^2q^2}{2},\qquad S=W(q;E)-Et.
$$

Writing $A=\sqrt{2E/(m\omega^2)}$, the HJ equation separates immediately:

$$
p=\frac{dW}{dq}=\pm m\omega\sqrt{A^2-q^2}.
$$

On the positive-momentum branch, choose the reference $W(0;E)=0$:

$$
W_+(q;E)=\frac{m\omega}{2}q\sqrt{A^2-q^2}+\frac{E}{\omega}\arcsin\left(\frac qA\right).
$$

The cancellation of the energy derivatives in this expression gives

$$
\frac{\partial W_+}{\partial E}=\frac1\omega\arcsin\left(\frac qA\right)=t+\beta_E.
$$

Thus, after continuing through successive branches,

$$
q(t)=A\sin\left(\omega(t+\beta_E)\right),\qquad p(t)=m\omega A\cos\left(\omega(t+\beta_E)\right).
$$

The principal branch of arcsine describes only a portion of the orbit. Energy selects the ellipse in phase space; $\beta_E$ selects the phase on that ellipse. The equilibrium $E=0$ is a separate limiting case.

## Pendulum: separation gives a quadrature, not necessarily an elementary function

For a pendulum of length $l$, mass $m$, and angle $\theta$, let $I=ml^2$ and $\omega_0=\sqrt{g/l}$:

$$
H=\frac{p_\theta^2}{2I}+mgl\left(1-\cos{\theta}\right),\qquad S=W(\theta;E)-Et.
$$

Separation gives

$$
W(\theta;E)=\pm\int^\theta\sqrt{2I\left[E-mgl(1-\cos{\vartheta})\right]}\,d\vartheta,
$$

and its energy derivative gives time along a momentum branch:

$$
t+\beta_E=\int^\theta\frac{I\,d\vartheta}{p_\theta(\vartheta;E)}.
$$

For librations, $0<E<2mgl$, put $k=\sin(\theta_{\max}/2)=\sqrt{E/(2mgl)}$. With a zero-angle crossing at $t=t_0$ and positive angular velocity,

$$
\sin\left(\frac{\theta(t)}2\right)=k\,\operatorname{sn}\left(\omega_0(t-t_0)\mid k^2\right),\qquad T(E)=\frac4{\omega_0}K(k^2).
$$

Here $K(M)=\int_0^{\pi/2}(1-M\sin^2u)^{-1/2}du$ and $\operatorname{sn}(u\mid M)$ use the **parameter** $M=k^2$, as do SciPy's elliptic functions. For libration, $\theta=2\arcsin(k\operatorname{sn})$ is continuous because $k<1$.

As $E\to0$, $T\to2\pi/\omega_0$, recovering the oscillator. At $E=2mgl$ the separatrix has infinite traversal time; the unstable equilibrium at the top is a separate stationary solution. For $E>2mgl$, the same HJ quadrature describes rotations, but the libration formula and its period must not be reused unchanged.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.special import ellipj, ellipk

# Dimensionless numerical check: m=l=g=1; libration starts at theta=0.
hj_times=np.linspace(0,16,2001)
hj_k=0.7
hj_sol=solve_ivp(lambda t,y:[y[1],-np.sin(y[0])],(0,16),[0,2*hj_k],t_eval=hj_times,rtol=1e-10,atol=1e-12)
assert hj_sol.success
hj_exact=2*np.arcsin(hj_k*ellipj(hj_times,hj_k**2)[0])
hj_error=np.max(np.abs(hj_sol.y[0]-hj_exact))
assert hj_error<1e-7
fig,hj_axes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
hj_axes[0].plot(hj_times,hj_sol.y[0],label='Hamilton equations')
hj_axes[0].plot(hj_times,hj_exact,'--',label='Separated HJ solution')
hj_axes[0].plot(hj_times,2*hj_k*np.sin(hj_times),':',label='Small-angle approximation')
hj_axes[0].set(xlabel='Dimensionless time',ylabel='Angle (rad)');hj_axes[0].legend()
hj_moduli=np.linspace(0,0.995,200)
hj_axes[1].plot(hj_moduli,4*ellipk(hj_moduli**2)/(2*np.pi))
hj_axes[1].set(xlabel='k = sin(theta_max/2)',ylabel='T / small-angle period')
plt.show()
print(f'Pendulum maximum angle difference: {hj_error:.3e} rad')
```

## Central fields: energy and angular momentum separate the motion

A nonradial central-force orbit lies in a fixed plane. In plane polar coordinates,

$$
H=\frac{p_r^2}{2m}+\frac{p_\varphi^2}{2mr^2}+V(r),\qquad S=-Et+\ell\varphi+W_r(r;E,\ell).
$$

The cyclic angle gives $p_\varphi=\ell$, the signed angular momentum perpendicular to the orbital plane. The radial equation becomes

$$
p_r=\pm\sqrt{2m\left[E-V(r)\right]-\frac{\ell^2}{r^2}},\qquad W_r=\int^r p_r(\rho;E,\ell)\,d\rho.
$$

The effective potential $V_{\rm eff}=V+\ell^2/(2mr^2)$ determines allowed radial intervals. Parameter derivatives recover both time and orbit:

$$
t+\beta_E=\int^r\frac{m\,d\rho}{p_r(\rho)},\qquad \varphi-\beta_\ell=\int^r\frac{\ell\,d\rho}{\rho^2p_r(\rho)}.
$$

These expressions apply branch by branch. A central potential always admits this reduction, although the quadratures need not be elementary and bounded orbits need not close. Circular orbits are limiting cases satisfying $E=V_{\rm eff}$ and $V'_{\rm eff}=0$; the turning-point quadrature is singular there and the circular solution is found directly. Radial motion $\ell=0$ is another special case.

### The additive structure in three dimensions

In spherical coordinates, one may instead retain the full separation:

$$
S=-Et+W_r(r)+W_\theta(\theta)+\ell_z\varphi,
$$

$$
\left(\frac{dW_\theta}{d\theta}\right)^2+\frac{\ell_z^2}{\sin^2\theta}=L^2,\qquad
\left(\frac{dW_r}{dr}\right)^2=2m\left[E-V(r)\right]-\frac{L^2}{r^2}.
$$

The separation constants are $E$, $L^2$ and $\ell_z$ on a regular patch. The planar reduction simply chooses axes adapted to the conserved angular momentum. Coordinate-axis singularities and degenerate motions need separate charts; they do not invalidate the local separation scheme.

## Kepler problem: evaluating the separated orbit

Let $V(r)=-\kappa/r$ with $\kappa>0$. For $\ell\ne0$,

$$
p_r^2=2mE+\frac{2m\kappa}{r}-\frac{\ell^2}{r^2}.
$$

In the angular quadrature, substitute $u=1/r$. The resulting quadratic under the square root integrates to a conic:

$$
r(\varphi)=\frac{b}{1+e\cos\left(\varphi-\varphi_0\right)},\qquad b=\frac{\ell^2}{m\kappa},\qquad e=\sqrt{1+\frac{2E\ell^2}{m\kappa^2}}.
$$

The parameter $b$ is the semilatus rectum (not a canonical momentum). Energy classifies the noncollision orbits: $E<0$ gives ellipses, $E=0$ parabolas and $E>0$ hyperbolas; $e=0$ is a circle. The orientation $\varphi_0$ is fixed by the integration constants.

For a bound ellipse,

$$
a=-\frac{\kappa}{2E},\qquad b=a(1-e^2),\qquad T=2\pi\sqrt{\frac{ma^3}{\kappa}}.
$$

The energy quadrature can be evaluated with eccentric anomaly $u$ (a new parameter, not an inverse radius in the next formulas):

$$
r=a(1-e\cos{u}),\qquad n(t-t_{\rm peri})=u-e\sin{u},\qquad n=\sqrt{\frac{\kappa}{ma^3}}.
$$

This is Kepler's equation. Solving for $u(t)$ supplies the time parametrization, which the conic equation alone does not provide. For $0<e<1$ the geometric relation is

$$
\tan\left(\frac{\varphi-\varphi_0}{2}\right)=\sqrt{\frac{1+e}{1-e}}\tan\left(\frac u2\right),
$$

with the orbital angle continued through the appropriate quadrants. The two elementary parameterizations arise from the same separated radial action, differentiated with respect to angular momentum or energy.

```python
# Compare the separated Kepler conic with Hamilton's equations (m=kappa=1).
hj_e=0.5;hj_a=1.5;hj_b=hj_a*(1-hj_e**2)
hj_rperi=hj_a*(1-hj_e)
hj_vperi=np.sqrt((1+hj_e)/(hj_a*(1-hj_e)))
hj_period=2*np.pi*hj_a**1.5

def hj_kepler_rhs(t,state):
    x,y,px,py=state
    radius=np.hypot(x,y)
    return [px,py,-x/radius**3,-y/radius**3]

hj_orbit=solve_ivp(hj_kepler_rhs,(0,hj_period),[hj_rperi,0,0,hj_vperi],rtol=1e-10,atol=1e-12,t_eval=np.linspace(0,hj_period,1601))
assert hj_orbit.success
hj_x,hj_y,hj_px,hj_py=hj_orbit.y
hj_radius=np.hypot(hj_x,hj_y)
hj_phi=np.unwrap(np.arctan2(hj_y,hj_x))
hj_radial_error=np.max(np.abs(hj_radius-hj_b/(1+hj_e*np.cos(hj_phi))))
assert hj_radial_error<1e-7
hj_energy=(hj_px**2+hj_py**2)/2-1/hj_radius
assert np.max(np.abs(hj_energy+1/(2*hj_a)))<1e-8
fig,hj_ax=plt.subplots(figsize=(6,5),layout='constrained')
hj_curve=np.linspace(0,2*np.pi,800);hj_rconic=hj_b/(1+hj_e*np.cos(hj_curve))
hj_ax.plot(hj_x,hj_y,label='Hamilton equations')
hj_ax.plot(hj_rconic*np.cos(hj_curve),hj_rconic*np.sin(hj_curve),'--',label='Separated HJ conic')
hj_ax.plot(0,0,'o',label='Force centre')
hj_ax.set(xlabel='x',ylabel='y',aspect='equal');hj_ax.legend();plt.show()
print(f'Kepler maximum radial difference: {hj_radial_error:.3e}')
```

### Accumulating the action along a Kepler characteristic

To check the characteristic equation for S itself, integrate an additional ODE together with Hamilton's equations. For a regular kinetic-plus-potential Hamiltonian,

$$
\dot S=p_i\dot q^i-H=\frac{p_x^2+p_y^2}{2m}+\frac{\kappa}{r}.
$$

For a bound ellipse starting at periapsis, the eccentric anomaly u gives an independent analytic result:

$$
S(t)-S(0)=\sqrt{m\kappa a}\left(\frac32u+\frac e2\sin{u}\right),\qquad
n t=u-e\sin{u}.
$$

After a complete orbit the phase-space point returns, but the accumulated principal action increases by $3\pi\sqrt{m\kappa a}$. This explicitly demonstrates that S is not a first integral; it must be interpreted along a continued branch. The notation e in this orbital formula means eccentricity, unlike the extended time-gradient component used in the general derivation above.

```python
from scipy.optimize import brentq

def hj_characteristic_with_action(t,state):
    x,y,px,py,action=state
    r=np.hypot(x,y)
    return [px,py,-x/r**3,-y/r**3,(px**2+py**2)/2+1/r]

hj_action_times=np.linspace(0,hj_period,801)
hj_action_sol=solve_ivp(hj_characteristic_with_action,(0,hj_period),[hj_rperi,0,0,hj_vperi,0],t_eval=hj_action_times,rtol=1e-10,atol=1e-12)
assert hj_action_sol.success
hj_mean=hj_action_times/hj_a**1.5
hj_anomaly=np.array([brentq(lambda u:u-hj_e*np.sin(u)-M,0,2*np.pi) if 0<M<2*np.pi else M for M in hj_mean])
hj_action_exact=np.sqrt(hj_a)*(1.5*hj_anomaly+0.5*hj_e*np.sin(hj_anomaly))
hj_action_error=np.max(np.abs(hj_action_sol.y[4]-hj_action_exact))
assert hj_action_error<1e-7
fig,hj_ax=plt.subplots(figsize=(8,3.5),layout='constrained')
hj_ax.plot(hj_action_times,hj_action_sol.y[4],label='Characteristic ODE for S')
hj_ax.plot(hj_action_times,hj_action_exact,'--',label='Separated Kepler result')
hj_ax.set(xlabel='Time',ylabel='S(t)-S(0)');hj_ax.legend();plt.show()
print(f'Characteristic action maximum difference: {hj_action_error:.3e}')
```

## A nontrivial orthogonal system: elliptic coordinates

The [generalized-coordinates lecture](https://colab.research.google.com/github/kulinskij-tech/ClassMech/blob/main/ClassMech_Python/classmech_gencoord.ipynb) introduces confocal ellipses and hyperbolas through elliptic coordinates. Restore a focal length $a>0$ (the earlier notebook uses $a=1$):

$$
x=a\cosh{u}\cos{v},\qquad y=a\sinh{u}\sin{v}.
$$

The foci are $(\pm a,0)$. Work locally with $u>0$ and an angular branch for $v$; the focal segment and coordinate degeneracies require separate patches. Differentiating the transformation gives

$$
ds^2=h^2(u,v)(du^2+dv^2),\qquad h^2=a^2\left(\sinh^2u+\sin^2v\right).
$$

This diagonal metric proves orthogonality. The equal scale factors give

$$
T=\frac{m h^2}{2}\left(\dot u^2+\dot v^2\right),\qquad
p_u=mh^2\dot u,\qquad p_v=mh^2\dot v,
$$

$$
H=\frac{p_u^2+p_v^2}{2m h^2}+V(u,v).
$$

As elsewhere, momenta are covector components: $p_x\,dx+p_y\,dy=p_u\,du+p_v\,dv$. The metric enters the **kinetic energy**, while the canonical brackets remain $\{u,p_u\}=\{v,p_v\}=1$, with zero mixed brackets. They are not metric components.

### Which potentials allow additive separation?

Try $S=-Et+U(u)+B(v)$ and multiply the stationary HJ equation by $h^2$:

$$
\frac{(U')^2}{2m}+\frac{(B')^2}{2m}+h^2V
=Ea^2\sinh^2u+Ea^2\sin^2v.
$$

All terms except $h^2V$ already separate. Therefore the locally general smooth potential compatible with this additive ansatz is

$$
\boxed{V(u,v)=\frac{F(u)+G(v)}{a^2\left(\sinh^2u+\sin^2v\right)}}.
$$

Equivalently, $\partial_u\partial_v(h^2V)=0$ on a regular coordinate patch. Integrating this mixed-derivative condition gives $h^2V=F(u)+G(v)$. The decomposition has the harmless freedom $F\mapsto F+c$, $G\mapsto G-c$, which shifts the separation constant. Here $F,G$ have dimensions of energy times length squared. They are arbitrary smooth one-variable functions locally; a globally smooth, single-valued physical potential can impose additional regularity and periodicity conditions.

Orthogonality alone does not guarantee separation. For a general orthogonal metric, $H=\sum_i p_i^2/(2m h_i^2)+V$ has no cross-momentum terms, but the coordinate dependence of the scale factors and the potential must still be compatible with separation. The displayed quotient is the required potential family for **this elliptic metric**, not a universal formula for all orthogonal systems.

### Separated equations, first integrals, and characteristics

Introduce the second essential constant C:

$$
\frac{(U')^2}{2m}+F(u)-Ea^2\sinh^2u=C,\qquad
\frac{(B')^2}{2m}+G(v)-Ea^2\sin^2v=-C.
$$

Thus the complete integral is obtained by two quadratures:

$$
U=\int^u\!\pm\sqrt{2m\left[C+Ea^2\sinh^2s-F(s)\right]}\,ds,
$$

$$
B=\int^v\!\pm\sqrt{2m\left[-C+Ea^2\sin^2s-G(s)\right]}\,ds.
$$

Allowed intervals require nonnegative radicands. As before, change branches at turning points. From $\partial_CS=\beta_C$ and $\partial_ES=\beta_E$,

$$
\beta_C=\int^u\frac{m\,ds}{p_u(s)}-\int^v\frac{m\,ds}{p_v(s)},
$$

$$
t+\beta_E=\int^u\frac{ma^2\sinh^2s}{p_u(s)}\,ds+\int^v\frac{ma^2\sin^2s}{p_v(s)}\,ds.
$$

These relations reconstruct the same Hamiltonian characteristics as direct integration. The unphysical parameter $\tau$, defined by $d\tau=dt/h^2$, makes this especially transparent:

$$
\frac{du}{d\tau}=\frac{p_u}{m},\qquad \frac{dv}{d\tau}=\frac{p_v}{m},\qquad
\frac{dt}{d\tau}=h^2.
$$

At fixed E, each momentum then depends only on its own coordinate and C. Physical time still couples the separated motions through the last equation. On the full phase space the second conserved function is

$$
C=\frac{p_u^2}{2m}+F(u)-a^2\sinh^2u\,H.
$$

Replacing H by a fixed energy E is justified only on that energy surface. The code below checks conservation along a numerical characteristic.

### Example A: the isotropic oscillator separates in elliptic coordinates too

For $V=m\omega^2(x^2+y^2)/2$, multiplication by the scale factor gives

$$
h^2V=\frac{m\omega^2a^4}{2}\left[\sinh^2u\cosh^2u+\sin^2v\cos^2v\right].
$$

Thus choose

$$
F(u)=\frac{m\omega^2a^4}{2}\sinh^2u\cosh^2u,\qquad
G(v)=\frac{m\omega^2a^4}{2}\sin^2v\cos^2v.
$$

The same oscillator admits Cartesian and elliptic separation. Different coordinate systems organize the constants differently; C here is not simply the Cartesian x-oscillator energy. We now integrate the elliptic-coordinate Hamiltonian and map the result back to Cartesian space. With $m=\omega=a=1$, the initial data correspond to $x(0)=2$, $y(0)=0$, $\dot x(0)=0$, $\dot y(0)=0.8$. The analytic characteristic is $x=2\cos t$, $y=0.8\sin t$ and remains away from the focal segment.

```python
# Elliptic-coordinate characteristic versus the known Cartesian oscillator.
# Dimensionless parameters: m=omega=a=1.
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def hj_elliptic_terms(u,v):
    su,cu,sv,cv=np.sinh(u),np.cosh(u),np.sin(v),np.cos(v)
    D=su**2+sv**2
    F=0.5*su**2*cu**2
    G=0.5*sv**2*cv**2
    Fu=su*cu*(su**2+cu**2)
    Gv=sv*cv*(cv**2-sv**2)
    return D,F,G,Fu,Gv

def hj_elliptic_rhs(t,state):
    u,v,pu,pv=state
    D,F,G,Fu,Gv=hj_elliptic_terms(u,v)
    H=(0.5*(pu**2+pv**2)+F+G)/D
    return [pu/D,pv/D,(2*H*np.sinh(u)*np.cosh(u)-Fu)/D,
            (2*H*np.sin(v)*np.cos(v)-Gv)/D]

hj_etimes=np.linspace(0,2*np.pi,1601)
hj_esol=solve_ivp(hj_elliptic_rhs,(0,2*np.pi),[np.arccosh(2),0,0,0.8*np.sqrt(3)],t_eval=hj_etimes,rtol=1e-10,atol=1e-12,max_step=0.025)
assert hj_esol.success
hj_u,hj_v,hj_pu,hj_pv=hj_esol.y
hj_ex=np.cosh(hj_u)*np.cos(hj_v)
hj_ey=np.sinh(hj_u)*np.sin(hj_v)
hj_eerror=np.max(np.hypot(hj_ex-2*np.cos(hj_etimes),hj_ey-0.8*np.sin(hj_etimes)))
hj_D,hj_F,hj_G,_,_=hj_elliptic_terms(hj_u,hj_v)
assert np.min(hj_D)>0.1
hj_eH=(0.5*(hj_pu**2+hj_pv**2)+hj_F+hj_G)/hj_D
hj_eC=0.5*hj_pu**2+hj_F-np.sinh(hj_u)**2*hj_eH
assert hj_eerror<1e-7
assert np.max(np.abs(hj_eH-hj_eH[0]))<1e-8
assert np.max(np.abs(hj_eC-hj_eC[0]))<1e-8
fig,hj_eaxes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
hj_gridv=np.linspace(0,2*np.pi,300)
for u0 in (0.4,0.7,1.0,1.3):
    hj_eaxes[0].plot(np.cosh(u0)*np.cos(hj_gridv),np.sinh(u0)*np.sin(hj_gridv),color='0.85',lw=0.7)
hj_eaxes[0].plot(hj_ex,hj_ey,label='Elliptic Hamiltonian characteristics')
hj_eaxes[0].plot(2*np.cos(hj_etimes),0.8*np.sin(hj_etimes),'--',label='Cartesian oscillator')
hj_eaxes[0].plot([-1,1],[0,0],'o',color='black',label='Foci')
hj_eaxes[0].set(xlabel='x',ylabel='y',aspect='equal');hj_eaxes[0].legend(fontsize=8)
hj_eaxes[1].plot(hj_etimes,hj_eH-hj_eH[0],label='H-H(0)')
hj_eaxes[1].plot(hj_etimes,hj_eC-hj_eC[0],label='C-C(0)')
hj_eaxes[1].set(xlabel='Time',ylabel='Absolute invariant drift');hj_eaxes[1].legend()
plt.show()
print(f'Elliptic oscillator maximum Cartesian difference: {hj_eerror:.3e}')
print(f'Separation invariant C(0)={hj_eC[0]:.6f}; maximum drift={np.max(np.abs(hj_eC-hj_eC[0])):.3e}')
```

### Example B: two fixed attractive centres

A genuinely noncentral member of the same family is Euler's two-centre potential. Let the centres at $(a,0)$ and $(-a,0)$ have strengths $\kappa_+>0$ and $\kappa_->0$, respectively. Their distances are

$$
r_+=a(\cosh u-\cos v),\qquad r_-=a(\cosh u+\cos v),\qquad r_+r_-=h^2.
$$

Consequently,

$$
V=-\frac{\kappa_+}{r_+}-\frac{\kappa_-}{r_-}
=\frac{-a(\kappa_++\kappa_-)\cosh u-a(\kappa_+-\kappa_-)\cos v}{h^2}.
$$

It separates with $F=-a(\kappa_++\kappa_-)\cosh u$ and $G=-a(\kappa_+-\kappa_-)\cos v$, using exactly the quadratures derived above. Ordinary angular momentum about the origin is generally not conserved, because the field is not central. The elliptic separation constant supplies the second integral instead. With one strength set to zero, this becomes a Kepler potential centred at the remaining focus. Collisions at either attracting centre require separate treatment.

## What additive separation accomplishes

For a sum of independent one-dimensional Hamiltonians, the separated constants are individual energies. For a central field, the constants come from angular symmetries and the radial energy equation. The pendulum shows that successful separation need not lead to elementary functions; the Kepler problem shows that orbit shape and time parametrization are different derivatives of the same generating function.

For periodic one-dimensional motion, a useful next step is the action variable

$$
J=\frac1{2\pi}\oint p\,dq,\qquad \frac{dJ}{dE}=\frac{T(E)}{2\pi}.
$$

The oscillator has $J=E/\omega$; the pendulum's nonconstant period means that energy is not linear in its action. The distinction between a local complete integral and a global action-angle description will matter near separatrices and in higher-dimensional systems.

## Canonical transformations

In Lagrangian mechanics, we have shown that the Euler–Lagrange equations are invariant under a change of generalized coordinates in configuration space:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right)-\frac{\partial L}{\partial q}=0\Leftrightarrow \frac{d}{dt}\left(\frac{\partial \widetilde{L}}{\partial \dot{Q}}\right)-\frac{\partial \widetilde{L}}{\partial Q}=0
$$

We can ask the same question about Hamilton’s equations of motion.

$$
\left\{\begin{matrix}\dot{q}=\frac{\partial H}{\partial p}\\ \dot{p}=-\frac{\partial H}{\partial q}\end{matrix}\right.\Leftrightarrow \left\{\begin{matrix}\dot{Q}=\frac{\partial \widetilde{H}}{\partial P}\\ \dot{P}=-\frac{\partial \widetilde{H}}{\partial Q}\end{matrix}\right.
$$

These equations follow from the variational principle:

$$
{\delta \int }_{{t}_{1}}^{{t}_{2}}\left(p dq-H dt\right)=0\Leftrightarrow {\delta \int }_{{t}_{1}}^{{t}_{2}}\left(P dQ-\widetilde{H} dt\right)=0\tag{1}
$$

The two variational principles yield the same equations of motion when their integrands differ by the total derivative of a generating function $F$:

$$
\left(p dq-H dt\right)-\left(P dQ-\widetilde{H} dt\right)=dF\Rightarrow p=\frac{\partial F}{\partial q},P=-\frac{\partial F}{\partial Q},\widetilde{H}-H=\frac{\partial F}{\partial t}\\ \forall F\left(q,Q,t\right)\tag{2}
$$

We can choose a different set of variables:

$$
\left(p dq-H dt\right)+Q dP-\widetilde{H} dt=d\Phi \Rightarrow p=\frac{\partial \Phi }{\partial q},Q=\frac{\partial \Phi }{\partial P},\widetilde{H}-H=\frac{\partial \Phi }{\partial t}\\ \forall \Phi \left(q,P,t\right)=F+PQ\tag{3}
$$

The Poisson bracket is also invariant under a canonical transformation:

$$
{\left\{{f}_{1},{f}_{2}\right\}}_{q,p}={\left\{{f}_{1},{f}_{2}\right\}}_{Q,P}
$$

The evolution of a state in phase space can itself be regarded as a canonical transformation:

$$
q\left({t}_{0}\right),p\left({t}_{0}\right)\to q\left({t}_{0}+\Delta t\right),p\left({t}_{0}+\Delta t\right)
$$

with the action $S\left(q\left(t_0\right),q\left(t_0+\Delta t\right)\right)$ as the generating function.

## Hamilton-Jacobi equation and separation of variables

Hamilton’s equations of motion can be viewed as a special case of the integration of a first-order partial differential equation:

$$
F\left(\mathbf{q},u,{\partial }_{\mathbf{q}}u\right)=0\tag{4}
$$

using the method of characteristics. A solution of equation (4) defines a surface in the extended $\left(\mathbf{q},u,\mathbf{p}\right)$ space of coordinates, function values, and gradients:

$$
F\left(\mathbf{q},u,\mathbf{p}\right)=0\tag{5}
$$

with the following consistency condition for a solution $u\left(\mathbf{q}\right)$:

$$
\dot{u}=\dot{\mathbf{q}}\cdot \mathbf{p}
$$

The characteristic vector field generates a flow that leaves the surface in equation (5) invariant:

$$
\frac{d}{dt}F\left(\mathbf{q},u,\mathbf{p}\right)=\dot{\mathbf{q}}{\partial }_{q}F+\dot{u}{\partial }_{u}F+\mathbf{\dot{\mathbf{p}}}{\partial }_{\mathbf{p}}F=0\\ \dot{\mathbf{q}}\cdot \left({\partial }_{\mathbf{q}}F+\mathbf{p}{\partial }_{u}F\right)+\mathbf{\dot{\mathbf{p}}}\cdot {\partial }_{\mathbf{p}}F=0
$$

This gives the characteristic equations:

$$
\dot{\mathbf{q}}={\partial }_{\mathbf{p}}F,\mathbf{\dot{\mathbf{p}}}=-\left({\partial }_{\mathbf{q}}F+\mathbf{p}{\partial }_{u}F\right)
$$

The characteristics of the Hamilton–Jacobi equation are precisely Hamilton’s equations of motion derived earlier. Integrating Hamilton’s equations is therefore equivalent to integrating the Hamilton–Jacobi equation. The latter can be easier when the Hamiltonian admits separation of variables. Suppose a coordinate $q_k$ separates from the others, so that the Hamiltonian can be written as

$$
H\left(f\left({q}_{k},{\partial }_{{q}_{k}}S\right),...\right)
$$

where the omitted terms are independent of $q_k$ and $\partial_{q_k}S$. The Hamilton–Jacobi equation can then be partially integrated with respect to $q_k$:

$$
S = {S}_{k}\left({q}_{k}\right)+{S}_{r}\left(\overline{q}\right)\Rightarrow H\left(f\left({q}_{k},{\partial }_{{q}_{k}}S\right),...\right)=E \Rightarrow  f\left({q}_{k},{\partial }_{{q}_{k}}S\right)={C}_{k} \\ H\left({C}_{k},\overline{q},\partial {S}_{r}\left(\overline{q}\right)\right)=E
$$

Here $\overline{q}$ denotes all coordinates except $q_k$, and $C_k$ is an integration constant. Let’s illustrate this.
