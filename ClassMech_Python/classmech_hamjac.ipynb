# Integration of Hamilton-Jacobi equation

## Canonical transformations

In Lagrangian mechanics, we have shown that the Euler–Lagrange equations are invariant under a change of generalized coordinates in configuration space:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right)-\frac{\partial L}{\partial q}=0\Leftrightarrow \frac{d}{dt}\left(\frac{\partial \widetilde{L}}{\partial \dot{Q}}\right)-\frac{\partial \widetilde{L}}{\partial Q}=0
$$

We can ask the same question about Hamilton’s equations of motion.

$$
\left\{\begin{matrix}\dot{q}=\frac{\partial H}{\partial p}\\ \dot{p}=-\frac{\partial H}{\partial q}\end{matrix}\right.\Leftrightarrow \left\{\begin{matrix}\dot{Q}=\frac{\partial \widetilde{H}}{\partial P}\\ \dot{P}=-\frac{\partial \widetilde{H}}{\partial Q}\end{matrix}\right.
$$

These equations follow from the variational principle:

$$
{\delta \int }_{{t}_{1}}^{{t}_{2}}\left(p dq-H dt\right)=0\Leftrightarrow {\delta \int }_{{t}_{1}}^{{t}_{2}}\left(P dQ-\widetilde{H} dt\right)=0\tag{1}
$$

The two variational principles yield the same equations of motion when their integrands differ by the total derivative of a generating function $F$:

$$
\left(p dq-H dt\right)-\left(P dQ-\widetilde{H} dt\right)=dF\Rightarrow p=\frac{\partial F}{\partial q},P=-\frac{\partial F}{\partial Q},\widetilde{H}-H=\frac{\partial F}{\partial t}\\ \forall F\left(q,Q,t\right)\tag{2}
$$

We can choose a different set of variables:

$$
\left(p dq-H dt\right)+Q dP-\widetilde{H} dt=d\Phi \Rightarrow p=\frac{\partial \Phi }{\partial q},Q=\frac{\partial \Phi }{\partial P},\widetilde{H}-H=\frac{\partial \Phi }{\partial t}\\ \forall \Phi \left(q,P,t\right)=F+PQ\tag{3}
$$

The Poisson bracket is also invariant under a canonical transformation:

$$
{\left\{{f}_{1},{f}_{2}\right\}}_{q,p}={\left\{{f}_{1},{f}_{2}\right\}}_{Q,P}
$$

The evolution of a state in phase space can itself be regarded as a canonical transformation:

$$
q\left({t}_{0}\right),p\left({t}_{0}\right)\to q\left({t}_{0}+\Delta t\right),p\left({t}_{0}+\Delta t\right)
$$

with the action $S\left(q\left(t_0\right),q\left(t_0+\Delta t\right)\right)$ as the generating function.

## Hamilton-Jacobi equation and separation of variables

Hamilton’s equations of motion can be viewed as a special case of the integration of a first-order partial differential equation:

$$
F\left(\mathbf{q},u,{\partial }_{\mathbf{q}}u\right)=0\tag{4}
$$

using the method of characteristics. A solution of equation (4) defines a surface in the extended $\left(\mathbf{q},u,\mathbf{p}\right)$ space of coordinates, function values, and gradients:

$$
F\left(\mathbf{q},u,\mathbf{p}\right)=0\tag{5}
$$

with the following consistency condition for a solution $u\left(\mathbf{q}\right)$:

$$
\dot{u}=\dot{\mathbf{q}}\cdot \mathbf{p}
$$

The characteristic vector field generates a flow that leaves the surface in equation (5) invariant:

$$
\frac{d}{dt}F\left(\mathbf{q},u,\mathbf{p}\right)=\dot{\mathbf{q}}{\partial }_{q}F+\dot{u}{\partial }_{u}F+\mathbf{\dot{\mathbf{p}}}{\partial }_{\mathbf{p}}F=0\\ \dot{\mathbf{q}}\cdot \left({\partial }_{\mathbf{q}}F+\mathbf{p}{\partial }_{u}F\right)+\mathbf{\dot{\mathbf{p}}}\cdot {\partial }_{\mathbf{p}}F=0
$$

This gives the characteristic equations:

$$
\dot{\mathbf{q}}={\partial }_{\mathbf{p}}F,\mathbf{\dot{\mathbf{p}}}=-\left({\partial }_{\mathbf{q}}F+\mathbf{p}{\partial }_{u}F\right)
$$

The characteristics of the Hamilton–Jacobi equation are precisely Hamilton’s equations of motion derived earlier. Integrating Hamilton’s equations is therefore equivalent to integrating the Hamilton–Jacobi equation. The latter can be easier when the Hamiltonian admits separation of variables. Suppose a coordinate $q_k$ separates from the others, so that the Hamiltonian can be written as

$$
H\left(f\left({q}_{k},{\partial }_{{q}_{k}}S\right),...\right)
$$

where the omitted terms are independent of $q_k$ and $\partial_{q_k}S$. The Hamilton–Jacobi equation can then be partially integrated with respect to $q_k$:

$$
S = {S}_{k}\left({q}_{k}\right)+{S}_{r}\left(\overline{q}\right)\Rightarrow H\left(f\left({q}_{k},{\partial }_{{q}_{k}}S\right),...\right)=E \Rightarrow  f\left({q}_{k},{\partial }_{{q}_{k}}S\right)={C}_{k} \\ H\left({C}_{k},\overline{q},\partial {S}_{r}\left(\overline{q}\right)\right)=E
$$
