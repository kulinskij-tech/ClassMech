In the Leidenfrost effect, a water drop placed on a hot surface can survive for minutes. Under certain circumstances such a drop develops oscillating star shapes. Induce different oscillatory modes and investigate them.



### Solution hint:

Equations of motion of a liquid

$$
\frac{\partial  \mathbf{v}}{\partial  t}+\left(\mathbf{v}\cdot \nabla \right)\mathbf{v}\mathbf{=}\mathbf{-}\frac{\nabla p}{\rho }+\mathbf{g}\\ \frac{\partial  \mathbf{\rho }}{\partial  t}+\nabla \left(\rho  \mathbf{v}\right)=0,\tag{1}
$$

flow is incompressible

$$
\rho =\text{const} \Rightarrow {\rm div} \mathbf{v}=0\tag{2}
$$

and we assume that it is potential

$$
{\rm rot} \mathbf{v}=0\tag{3}
$$

So that as usual the potential $\psi $ of the velocity field can be introduced and (3) is as follows

$$
\Delta \psi =0\tag{4}
$$

The shape of the drop is naturally described in polar coordinates (center $r=0$ is the center of mass)

$$
r\left(\phi ,t\right)\tag{5}
$$

Bernoulli equation (we put density =1):

$$
\frac{\partial \psi }{\partial t}+\frac{1}{2}{\left\lvert \nabla \psi \right\rvert }^{2}+p=0\tag{6}
$$

Boundary conditions

$$
\frac{\partial \psi }{\partial t}+\frac{1}{2}{\left\lvert \nabla \psi \right\rvert }^{2}+\sigma  \left(\kappa -{\kappa }_{0}\right)=0\tag{7}
$$

Obvious boundary condition at the surface:

$$
\frac{\partial  r}{\partial  t}={v}_{r}={\left(\frac{\partial  \psi }{\partial  \rho }\right)}_{\rho =r\left(φ,t\right)}\tag{8}
$$
