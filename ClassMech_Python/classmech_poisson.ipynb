HAMILTONIAN MECHANICS: Hamiltonian mechanics is geometry in phase space. Phase space has the structure of a symplectic manifold. The group of symplectic diffeomorphisms acts on a symplectic manifold. The basic concepts and theorems of Hamiltonian mechanics (even if they are formulated in terms of local symplectic coordinates) are invariant under this group (and under a broader group of transformations involving time as well). A Hamiltonian mechanical system is specified by an even-dimensional manifold (“phase space”), a symplectic structure on it (the “Poincaré integral invariant”), and a function on it (the “Hamiltonian function”). Every one-parameter group of symplectic diffeomorphisms of phase space that preserves the Hamiltonian function is associated with a first integral of the equations of motion. Lagrangian mechanics is included in Hamiltonian mechanics as a special case (in this case, phase space is the cotangent bundle of configuration space, and the Hamiltonian function is the Legendre transformation of the Lagrangian function). The Hamiltonian point of view allows us to fully analyze a number of problems in mechanics that cannot be solved by other means (for example, the problem of attraction by two fixed centers and the problem of geodesics on a triaxial ellipsoid). The Hamiltonian point of view is of even greater importance for approximate methods of perturbation theory (celestial mechanics), for understanding the general nature of motion in complex mechanical systems (ergodic theory, statistical mechanics), and in connection with other branches of mathematical physics (optics, quantum mechanics, etc.).

DIFFERENTIAL FORMS: Differential exterior forms arise when generalizing to the multidimensional case such concepts as the work of a field along a path and the flux of a fluid through a surface. Hamiltonian mechanics cannot be understood without differential forms. The facts about differential forms that we need are exterior multiplication, exterior differentiation, integration, and Stokes’ formula.

The Hamiltonian formulation of mechanics uses the coordinate–momentum phase space $\left(q,p\right)$, of even dimension $2n$. Each physical quantity is a function of the phase-space point and time, $f\left(q,p,t\right)$. As the state evolves, the change in $f$ is:

$$
\frac{df\left(q,p,t\right)}{dt}=\frac{\partial f}{\partial  t}+\dot{q}\frac{\partial f}{\partial  q}+\dot{p}\frac{\partial f}{\partial  p}\tag{1}
$$

Summation over all coordinates is understood here and in subsequent formulas. Using Hamilton’s equations of motion gives:

$$
\frac{df\left(q,p,t\right)}{dt}=\frac{\partial f}{\partial  t}+\frac{\partial f}{\partial  q}\frac{\partial H}{\partial  p}-\frac{\partial f}{\partial  p}\frac{\partial H}{\partial  q}\\ \left\{f,H\right\}\overset{def}{=}\frac{\partial f}{\partial  q}\frac{\partial H}{\partial  p}-\frac{\partial H}{\partial  q}\frac{\partial f}{\partial  p}=-\left\{H,f\right\}\tag{2}
$$

This construction is called the Poisson bracket. Hamilton’s equations can be written in terms of it:

$$
\frac{dq}{dt}=\left\{q,H\right\},\frac{dp}{dt}=\left\{p,H\right\}\\ \left\{{q}^{i},{p}_{j}\right\}={\delta }_{j}^{i}\tag{3}
$$

A coordinate and momentum satisfying $\left\{q,p\right\}=1$ are called canonically conjugate.

Functions satisfying

$$
\frac{df\left(q,p,t\right)}{dt}=0\tag{4}
$$

are conserved quantities, or integrals of motion. If $f$ has no explicit time dependence, its conservation is equivalent to

$$
\left\{f,H\right\}=0\tag{5}
$$

The Poisson bracket can be defined for any pair of functions:

$$
\left\{g,f\right\}={∑}_{i}\left(\frac{∂g}{∂{q}^{i}}\frac{∂f}{∂{p}_{i}}−\frac{∂g}{∂{p}_{i}}\frac{∂f}{∂{q}^{i}}\right)\tag{6}
$$

Direct calculation gives the Jacobi identity:

$$
\left\{{f}_{1},\left\{{f}_{2},{f}_{3}\right\}\right\}+\left\{{f}_{3},\left\{{f}_{1},{f}_{2}\right\}\right\}+\left\{{f}_{2},\left\{{f}_{3},{f}_{1}\right\}\right\}=0\tag{7}
$$

It follows that the Poisson bracket of two conserved quantities is also conserved:

$$
\left\{H,\left\{{f}_{1},{f}_{2}\right\}\right\}+\left\{{f}_{2},\left\{H,{f}_{1}\right\}\right\}+\left\{{f}_{1},\left\{{f}_{2},H\right\}\right\}=\left\{H,\left\{{f}_{1},{f}_{2}\right\}\right\}=0\tag{8}
$$

```python
import sympy as sp
from IPython.display import display
pos = sp.Matrix(sp.symbols('x_1 x_2 x_3', real=True))
imp = sp.Matrix(sp.symbols('p_1 p_2 p_3', real=True))
mom = pos.cross(imp)

def poiss(a, b):
    if isinstance(a, sp.MatrixBase):
        return a.applyfunc(lambda entry: poiss(entry, b))
    if isinstance(b, sp.MatrixBase):
        return b.applyfunc(lambda entry: poiss(a, entry))
    return sp.simplify(sum(sp.diff(a,q)*sp.diff(b,p)-sp.diff(b,q)*sp.diff(a,p)
                           for q,p in zip(pos,imp)))
```

## Canonical transformations

In Lagrangian mechanics we have shown that Euler-Lagrange equation is invariant wrt to the choice of generalized coordinates in configurational space:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right)-\frac{\partial L}{\partial q}=0\Leftrightarrow \frac{d}{dt}\left(\frac{\partial \widetilde{L}}{\partial \dot{Q}}\right)-\frac{\partial \widetilde{L}}{\partial Q}=0
$$

We can ask the same question about Hamilton’s equations of motion.

$$
\left\{\begin{matrix}\dot{q}=\frac{\partial H}{\partial p}\\ \dot{p}=-\frac{\partial H}{\partial q}\end{matrix}\right.\Leftrightarrow \left\{\begin{matrix}\dot{Q}=\frac{\partial \widetilde{H}}{\partial P}\\ \dot{P}=-\frac{\partial \widetilde{H}}{\partial Q}\end{matrix}\right.
$$

These equations follow from the variational principle:

$$
{\delta \int }_{{t}_{1}}^{{t}_{2}}\left(p dq-H dt\right)=0\Leftrightarrow {\delta \int }_{{t}_{1}}^{{t}_{2}}\left(P dQ-\widetilde{H} dt\right)=0\tag{9}
$$

The two variational principles yield the same extremals only if their integrands differ by the total derivative of a generating function $F$:

$$
\left(p dq-H dt\right)-\left(P dQ-\widetilde{H} dt\right)=dF\Rightarrow p=\frac{\partial F}{\partial q},P=-\frac{\partial F}{\partial Q},\widetilde{H}-H=\frac{\partial F}{\partial t}\\ \forall F\left(q,Q,t\right)\tag{10}
$$

We can choose a different set of variables:

$$
\left(p dq-H dt\right)+Q dP-\widetilde{H} dt=d\Phi \Rightarrow p=\frac{\partial \Phi }{\partial q},Q=\frac{\partial \Phi }{\partial P},\widetilde{H}-H=\frac{\partial \Phi }{\partial t}\\ \forall \Phi \left(q,P,t\right)=F+PQ\tag{11}
$$

The Poisson bracket is also invariant under a canonical transformation:

$$
{\left\{{f}_{1},{f}_{2}\right\}}_{q,p}={\left\{{f}_{1},{f}_{2}\right\}}_{Q,P}
$$

The evolution of a state in phase space can itself be regarded as a canonical transformation:

$$
q\left({t}_{0}\right),p\left({t}_{0}\right)\to q\left({t}_{0}+\Delta t\right),p\left({t}_{0}+\Delta t\right)
$$

with the action $S\left(q\left(t_0\right),q\left(t_0+\Delta t\right)\right)$ as the generating function.

## Mathematical structures emerging from Hamiltonian mechanics

The mechanical material above has already introduced observables, their Poisson brackets, conserved quantities, angular momentum, canonical transformations, and Hamiltonian evolution. We now unfold the mathematical structures behind these constructions. Hamiltonian motion gives a vector field on phase space; canonical transformations suggest continuous groups; and brackets of observables reveal an algebra of generators. Starting from these physical examples, the following section develops the alphabet of vector fields, Lie brackets, Lie algebras, and Lie groups.

### Vector fields act on functions

A smooth vector field assigns a tangent vector at each point. In coordinates, regard it as a directional-derivative operator acting on a smooth scalar function h. Repeated coordinate indices are summed:

$$
X={X}^{i}{\partial }_{i},    Xh={X}^{i}\frac{\partial h}{\partial {x}^{i}},    \frac{d{x}^{i}}{ds}={X}^{i}\left(x\right)
$$

The last equation defines integral curves and the local flow of X. A flow gives a family of small transformations; it may exist only for a finite parameter interval. Vector fields can be added and multiplied by real constants. Their composition as differential operators is usually second order, but its antisymmetric part is again first order.

$$
\left[X,Y\right]h=X\left(Yh\right)−Y\left(Xh\right)    {\left[X,Y\right]}^{i}={X}^{j}{∂}_{j}{Y}^{i}−{Y}^{j}{∂}_{j}{X}^{i}
$$

The second derivatives of h cancel because mixed partial derivatives commute. Thus [X,Y] is a vector field. This is the Lie bracket. It compares differentiation along the two fields, including how their components vary from point to point.

### A small example: translation and shear

$$
X={\partial }_{x}    Y=x{\partial }_{y}    \left[X,Y\right]={\partial }_{y}
$$

The X flow translates x; the Y flow shears y by an amount proportional to x. Starting at (x,y), first apply X for parameter a and then Y for parameter b: the result is (x+a,y+b(x+a)). Reversing the order gives (x+a,y+bx). Their difference is (0,ab). The nonzero bracket identifies precisely this leading displacement. Constant translations in x and y commute, so their bracket is zero.

### Definition of a Lie algebra

A real Lie algebra is a real vector space equipped with a bilinear bracket satisfying antisymmetry and the Jacobi identity. The bracket need not be associative. Smooth vector fields, with the bracket above, form an infinite-dimensional Lie algebra. A finite-dimensional family closed under brackets gives a finite-dimensional Lie subalgebra.

$$
\left[X,Y\right]=-\left[Y,X\right]    \\ \left[X,\left[Y,Z\right]\right] + \left[Y,\left[Z,X\right]\right] + \left[Z,\left[X,Y\right]\right]=0 - Jacobi identity
$$

For a basis of generators, closure is expressed by structure constants. These depend on the chosen basis; Jacobi restricts which constants are possible. 

$$
\left[{T}_{a},{T}_{b}\right]={{c}_{ab}}^{c}{T}_{c}
$$

An algebra is called abelian when every bracket vanishes: $\left[{T}_{a},{T}_{b}\right]=$0.

### Lie groups describe finite transformations

A Lie group is a group that is also a smooth manifold, with smooth multiplication and inversion. Group elements represent finite transformations; its Lie algebra is the tangent space at the identity, with the bracket obtained from left-invariant vector fields. For matrix groups this bracket is the matrix commutator. Do not confuse a group element with a tangent generator.

$$
\left[A,B\right]=AB-BA    g\left(s\right)={e}^{sA}    g\left(s+t\right)=g\left(s\right)g\left(t\right)
$$

The exponential connects a generator to its one-parameter subgroup. It describes the group near the identity; one should not assume that every element of every Lie group is a single exponential. A Lie algebra captures local group structure, not all global topology. For example, the additive group of real numbers and the circle group have the same one-dimensional abelian Lie algebra, but only the circle has periodic group parameters.

Examples: translations form the additive group of real coordinate vectors and have an abelian Lie algebra. Planar rotations form SO(2), whose one-generator algebra is also abelian. Spatial rotations form SO(3), a nonabelian group. Its generators are real antisymmetric 3 by 3 matrices; the commutator of rotations about two axes produces the generator about the third.

$$
\left[{J}_{x},{J}_{y}\right]={J}_{z}    \left[{J}_{y},{J}_{z}\right]={J}_{x}    \left[{J}_{z},{J}_{x}\right]={J}_{y}
$$

### Poisson brackets: a Lie algebra of observables

For canonical coordinates, use the following convention throughout this lecture. It gives $\left\{{q}^{i},{p}_{j}\right\}={\delta }_{j}^{i}$ and reproduces Hamilton’s equations. Observables form a Lie algebra under the Poisson bracket; ordinary multiplication also obeys the Leibniz rule, so together these operations define a Poisson algebra.

$$
\left\{f,g\right\}={\sum }_{i}\left(\frac{\partial f}{\partial {q}^{i}}\frac{\partial g}{\partial {p}_{i}}-\frac{\partial f}{\partial {p}_{i}}\frac{\partial g}{\partial {q}^{i}}\right)
$$

The index positions distinguish configuration coordinates ${q}^{i}$ from the covector components ${p}_{i}$ on the cotangent bundle T*Q. The Kronecker delta is the identity pairing between coordinate directions and covectors; it is not a metric. No index is raised or lowered by a metric in the definition of the Poisson bracket. The coordinate functions ${q}^{i}$ themselves are not assumed to be components of a vector under a nonlinear coordinate change.

The canonical Poisson structure is an antisymmetric bivector. It acts on the differentials of observables to give their bracket. Its canonical expression and fundamental brackets are:

$$
\Pi ={\sum }_{i}\frac{\partial }{\partial {q}^{i}}\land \frac{\partial }{\partial {p}_{i}}    \left\{f,g\right\}=\Pi \left(df,dg\right)
$$

$$
\left\{{q}^{i},{p}_{j}\right\}={\delta }_{j}^{i}    \left\{{q}^{i},{q}^{j}\right\}=0    \left\{{p}_{i},{p}_{j}\right\}=0
$$

An observable f generates a Hamiltonian vector field. Constants generate the zero field, so distinct observables differing by a constant give the same infinitesimal transformation. The following sign is important: with ${X}_{f}$ acting on h as {h,f}, the map from observables to vector fields reverses the bracket sign. Jacobi gives this relation directly. Choosing instead ${D}_{f}h=\left\{f,h\right\}$  gives a bracket-preserving representation, but ${D}_{H}$ is minus the physical evolution operator.

$$
{X}_{f}={\sum }_{i}\left(\frac{\partial f}{\partial {p}_{i}}\frac{\partial }{\partial {q}^{i}}-\frac{\partial f}{\partial {q}^{i}}\frac{\partial }{\partial {p}_{i}}\right)    {X}_{f}h=\left\{h,f\right\}    \left[{X}_{f},{X}_{g}\right]=-{X}_{\left\{f,g\right\}}
$$

For one canonical pair, p generates translations of q, while q generates negative translations of p. Their Poisson bracket is the constant 1, but their vector fields commute because a constant has zero Hamiltonian vector field. This is a useful check on the distinction between observables and their generators.

$$
{X}_{p}={\partial }_{q}    {X}_{q}=-{\partial }_{p}    \left\{q,p\right\}=1    \left[{X}_{q},{X}_{p}\right]=0
$$

Angular momentum provides the rotational example. Its components close under the same structure constants as the matrix rotation algebra. The generator ${L}_{z}$ rotates both position and momentum about the z axis. The finite canonical rotation is obtained by following its Hamiltonian flow. The sign relation above must be retained when comparing the brackets of these physical vector fields with the matrix commutators.

$$
\left\{{L}_{x},{L}_{y}\right\}={L}_{z}    {X}_{{L}_{z}}=-y{\partial }_{x}+x{\partial }_{y}-{p}_{y}{\partial }_{{p}_{x}}+{p}_{x}{\partial }_{{p}_{y}}
$$

If two observables have no explicit time dependence and both commute with H, Jacobi implies that their Poisson bracket also commutes with H. Conserved observables therefore form a Lie subalgebra. This explains the algebraic closure encountered in the angular-momentum and Kepler examples above. Finite-dimensional closure with constant structure coefficients needs to be checked; it does not follow merely from having conserved quantities.

```python
# Independent checks of the introductory examples.
import sympy as sp
lie_x,lie_y,lie_q,lie_p=sp.symbols('lie_x lie_y lie_q lie_p', real=True)
def lie_bracket_components(A,B,coordinates):
    return sp.simplify(B.jacobian(coordinates)*A-A.jacobian(coordinates)*B)
lie_X=sp.Matrix([1,0]); lie_Y=sp.Matrix([0,lie_x])
assert lie_bracket_components(lie_X,lie_Y,(lie_x,lie_y))==sp.Matrix([0,1])
lie_Jx=sp.Matrix([[0,0,0],[0,0,-1],[0,1,0]])
lie_Jy=sp.Matrix([[0,0,1],[0,0,0],[-1,0,0]])
lie_Jz=sp.Matrix([[0,-1,0],[1,0,0],[0,0,0]])
assert lie_Jx*lie_Jy-lie_Jy*lie_Jx==lie_Jz
def lie_pb(f,g):
    return sp.diff(f,lie_q)*sp.diff(g,lie_p)-sp.diff(f,lie_p)*sp.diff(g,lie_q)
def lie_hamiltonian_field(f):
    return sp.Matrix([sp.diff(f,lie_p),-sp.diff(f,lie_q)])
lie_f=lie_q**2/2; lie_g=lie_p**2/2
assert lie_pb(lie_f,lie_g)==lie_q*lie_p
assert lie_bracket_components(lie_hamiltonian_field(lie_f),lie_hamiltonian_field(lie_g),(lie_q,lie_p))==-lie_hamiltonian_field(lie_pb(lie_f,lie_g))
print('Verified: vector-field bracket, rotation matrix commutator, and Hamiltonian sign convention.')
```
