# Stationary Action Principle

Action principles lie at the heart of fundamental physics, from classical mechanics through quantum mechanics, particle physics, and general relativity.

## Lagrange function (Lagrangian mechanics)

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def animate_parameter(draw,minimum,maximum,value,step,label):
    draw(value)
    slider=widgets.FloatSlider(value=value,min=minimum,max=maximum,step=step,
                               description=label,continuous_update=False)
    play=widgets.Play(value=0,min=0,max=int(round((maximum-minimum)/step)),step=1,interval=120,repeat=False)
    def advance(change):
        slider.value=min(maximum,minimum+change['new']*step)
    play.observe(advance,names='value')
    view=widgets.interactive_output(lambda parameter: draw(parameter),{'parameter':slider})
    display(widgets.VBox([widgets.HBox([play,slider]),view]))
    return slider,play,view
```

By the law of inertia, free motion follows a straight line—the shortest path between two points in Euclidean space.

$$
\mathbf{a}\mathbf{=}\frac{d\mathbf{v}}{dt}=0\Rightarrow {\ddot{q}}^{i}+{\Gamma }_{mk}^{i}{\dot{q}}^{m}{\dot{q}}^{k}=0\tag{1}
$$

We can therefore express the free-trajectory problem geometrically as finding the shortest path between the initial and final points.

$$
\text{Length} = {\int }_{\text{start}}^{\text{end}}ds={\int }_{{t}_{0}}^{t}\sqrt{{g}_{mn}{\dot{q}}^{m}{\dot{q}}^{n}}dt\tag{2}
$$

The calculus of variations studies extrema of functionals. A particularly important class has the form

$$
S = {\int }_{{t}_{1}}^{{t}_{2}}L\left(t,q,\dot{q}\right)dt\tag{3}
$$

For a stationary path, a small variation of the trajectory must satisfy

$$
0=\delta S={\int }_{{t}_{1}}^{{t}_{2}}\left(L\left(t,q+\delta q,\dot{q}+\delta \dot{q}\right)-L\left(t,q,\dot{q}\right)\right)dt={\int }_{{t}_{1}}^{{t}_{2}}\left(\frac{\partial L}{\partial q}\delta q+\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\delta q\right)-\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right)\delta q\right)dt\tag{4}
$$

which gives the second-order Euler–Lagrange equations:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial {\dot{q}}^{i}}\right)-\frac{\partial L}{\partial {q}^{i}}=0, i=1,\ldots ,N\tag{5}
$$

These can be interpreted as equations of motion in mechanics, electrodynamics, and other physical theories. For the length functional, we obtain:

$$
\frac{d}{dt}\left(\frac{{g}_{in}{\dot{q}}^{n}}{\sqrt{{g}_{mn}{\dot{q}}^{m}{\dot{q}}^{n}}}\right)-\frac{1}{2\sqrt{{g}_{mn}{\dot{q}}^{m}{\dot{q}}^{n}}}\frac{\partial {g}_{mn}}{\partial {q}^{i}}{\dot{q}}^{m}{\dot{q}}^{n}=0, i=1,\ldots ,N\tag{6}
$$

$$
\frac{\partial  {g}_{mn}}{\partial  {q}^{i}}={g}_{mk}{\Gamma }_{in}^{k}+{g}_{nk}{\Gamma }_{im}^{k}\tag{7}
$$

$$
\frac{d}{dt}\left(\frac{{g}_{in}{\dot{q}}^{n}}{\sqrt{{g}_{mn}{\dot{q}}^{m}{\dot{q}}^{n}}}\right)-\frac{1}{\sqrt{{g}_{mn}{\dot{q}}^{m}{\dot{q}}^{n}}}{g}_{mk}{\Gamma }_{in}^{k}{\dot{q}}^{m}{\dot{q}}^{n}=0\tag{8}
$$

If $t$ is physical time, free motion has constant speed $v=\sqrt{g_{mn}\dot q^m\dot q^n}$. Thus $s=vt$ is an arc-length parameter, and

$$
0=\frac{d}{dt}\left({g}_{in}{\dot{q}}^{n}\right)-{g}_{mk}{\Gamma }_{in}^{k}{\dot{q}}^{m}{\dot{q}}^{n}={g}_{in}{\ddot{q}}^{n}+\frac{\partial {g}_{in}}{\partial {q}^{m}}{\dot{q}}^{n}{\dot{q}}^{m}-{g}_{mk}{\Gamma }_{in}^{k}{\dot{q}}^{m}{\dot{q}}^{n}={g}_{in}{\ddot{q}}^{n}+\left({g}_{ik}{\Gamma }_{nm}^{k}+{g}_{mk}{\Gamma }_{in}^{k}\right){\dot{q}}^{n}{\dot{q}}^{m}-{g}_{mk}{\Gamma }_{in}^{k}{\dot{q}}^{m}{\dot{q}}^{n}={g}_{in}{\ddot{q}}^{n}+{g}_{ik}{\Gamma }_{mn}^{k}{\dot{q}}^{m}{\dot{q}}^{n}\tag{9}
$$

Finally we get

$$
\frac{{d }^{2}{q}^{i}}{{ds}^{2}}+{\Gamma }_{mn}^{i}\frac{d {q}^{m}}{ds}\frac{d {q}^{n}}{ds}=0\tag{10}
$$

With $s=vt$, this agrees with the kinematic expression for acceleration. In a general non-Euclidean geometry, the resulting equations define geodesics. The Christoffel symbols enter through the covariant derivative:

$$
{\nabla }_{{\mathbf{e}}_{i}}{\mathbf{e}}_{j}={\Gamma }_{ij}^{k}{\mathbf{e}}_{k}\tag{11}
$$

A geodesic transports its tangent vector $\mathbf{v}$ parallel to itself:

$$
{\nabla }_{\mathbf{v}}\mathbf{v}=0\tag{12}
$$

This gives the same geodesic equation.

Newton’s second law for $N$ particles in a potential field is

$$
{m}_{i}\mathbf{ }\frac{d {v}_{i}}{dt}\mathbf{=}-{\nabla }_{i}U, i=1,...,N\tag{13}
$$

This has the Euler–Lagrange form if the mechanical action uses

$$
S = {\int }_{{t}_{1}}^{{t}_{2}}L\left(t,q,\dot{q}\right)dt, L = \underset{i}{\sum }{m}_{i}\frac{{\mathbf{v}}_{i}^{2}}{2}-U\left({\mathbf{r}}_{1},...,{\mathbf{r}}_{N}\right)\tag{14}
$$

This is Hamilton’s variational principle. For $N$ particles in $d$ dimensions with $n$ independent geometric constraints, the number of degrees of freedom is $f=dN-n$. Choose suitable generalized coordinates $q_1,\ldots,q_f$. The Lagrangian for a given set of equations of motion is not unique: the function

$$
\widetilde{L}\left(t,q,\dot{q}\right)\to \lambda  L\left(t,q,\dot{q}\right)+\frac{d f\left(t,q,\dot{q}\right)}{dt}\tag{15}
$$

gives the same Euler–Lagrange equations as $L$.

By comparison with Newton’s second law, the quantity

$$
\left(\frac{\partial L}{\partial {\dot{q}}^{i}}\right)={p}_{i}\tag{16}
$$

is called the generalized momentum conjugate to $q^i$.

### Solution

In polar coordinates the length functional is

$$
\text{Length} = {\int }_{A}^{B}\sqrt{{\left(\frac{d\rho }{ds}\right)}^{2}+{\rho }^{2}{\left(\frac{d\varphi }{ds}\right)}^{2}}ds\tag{17}
$$

therefore

$$
\frac{{d}^{2}\rho }{{ds}^{2}}-2\rho {\left(\frac{d\varphi }{ds}\right)}^{2}=0, \frac{d}{ds}\left({\rho }^{2}\frac{d\varphi }{ds}\right)=0 \Rightarrow {\rho }^{2}\frac{d\varphi }{ds}=\frac{{c}_{1}}{\sqrt{2}}\tag{18}
$$

When $c_1=0$, the solutions are straight lines through the origin. The equations are invariant under the affine rescaling $s\mapsto as+b$.

$$
\varphi ={\varphi }_{0},\rho =s\tag{19}
$$

The equation for $\rho\left(s\right)$ can be integrated by multiplying by $\dot\rho\left(s\right)$ and changing variables:

$$
\frac{{d}^{2}\rho }{{ds}^{2}}=\frac{1}{{\rho }^{3}}{c}_{1}^{2} \Rightarrow \frac{d{\dot{\rho }}^{2}}{ds}=-{c}_{1}^{2}\frac{d}{ds}\left(\frac{1}{{\rho }^{2}}\right)\Rightarrow {\dot{\rho }}^{2}={c}_{2}-\frac{{c}_{1}^{2}}{{\rho }^{2}}\Rightarrow {\dot{\rho }}^{2}=1-{\left(\frac{{\rho }_{0}}{\rho }\right)}^{2}\Rightarrow \rho \left(s\right) = \sqrt{{\rho }_{0}^{2}+{s}^{2}}\tag{20}
$$

```python
s=sp.symbols('s',real=True)
rho_solution=sp.sqrt(1+s*s)
display(sp.Eq(sp.Function('rho')(s),rho_solution))
display(sp.simplify(sp.diff(rho_solution,s)**2-(1-1/rho_solution**2)))
# The squared first-order ODE also permits rho=1, and delayed departures:
# its squared form alone does not select the nonconstant geodesic branch.
display(sp.Eq(sp.Function('rho')(s),sp.Integer(1)))
```

Integrating the equation for $\varphi\left(s\right)$ then gives the expected geometry: straight lines.

The brachistochrone problem, another historical motivation for the calculus of variations, asks for the curve of fastest descent between points A and B.

```python
def bezier(t,px=0.5,py=-0.25):
    t=np.asarray(t)
    return (1-t)**2*np.array([[0.0],[1.0]])+2*(1-t)*t*np.array([[px],[py]])+t*t*np.array([[1.0],[0.0]])

def show_descent(time=0.0,px=0.5,py=-0.25):
    line=bezier(np.linspace(0,1,300),px,py); point=bezier(np.array([time]),px,py)[:,0]
    fig,ax=plt.subplots(figsize=(5,5),layout='constrained')
    ax.plot(*line); ax.plot([0,1],[1,0],'ko'); ax.plot(*point,'ro')
    ax.annotate('mg',xy=point+[0,-1/3],xytext=point,arrowprops={'arrowstyle':'->','color':'red'})
    ax.text(0,1.1,'A'); ax.text(1.05,0,'B')
    ax.set(xlim=(-0.1,1.15),ylim=(-0.5,1.3),aspect='equal'); plt.show()
show_descent()
bezier_controls=widgets.interactive(show_descent,time=widgets.FloatSlider(min=0,max=1,step=0.01),
                                    px=widgets.FloatSlider(min=-0.1,max=1.1,value=0.5,step=0.05),
                                    py=widgets.FloatSlider(min=-0.5,max=1.3,value=-0.25,step=0.05))
display(bezier_controls)
```

Suppose the curve is given by y(x). The solution is to minimize the functional of time

$$
\text{Time} = {\int }_{A}^{B}\frac{ds}{v}={\int }_{A}^{B}\frac{ds}{\sqrt{\frac{2}{m}\left(E-mgy\right)}}\tag{21}
$$

After rescaling the variables we have:

$$
\text{Time} = {\int }_{A}^{B}\frac{ds}{v}={\int }_{{x}_{A}}^{{x}_{B}}\sqrt{\frac{1+{\left(\frac{dy}{dx}\right)}^{2}}{1- y}}dx\tag{22}
$$

The Euler–Lagrange equation for $L=\sqrt{\left(1+\left(y\prime\right)^2\right)/\left(1-y\right)}$, which is proportional to $1/v_x$, is

$$
\frac{d}{dx}\left(\frac{\partial L}{\partial y'}\right)-\frac{\partial L}{\partial y}=0\tag{23}
$$

Since $L$ has no explicit dependence on $x$, the equation reduces to first order:

$$
y'\frac{d}{dx}\left(\frac{\partial L}{\partial y'}\right)-y'\frac{\partial L}{\partial y}=\frac{d}{dx}\left(y'\frac{\partial L}{\partial y'}\right)-\left(y'\frac{\partial L}{\partial y}+y''\frac{\partial L}{\partial y'}\right)=\\ \frac{d}{dx}\left(y'\frac{\partial L}{\partial y'}-L\right)=0\Rightarrow y'\frac{\partial L}{\partial y'}-L=\text{const}\\ -\frac{1}{\sqrt{\left(1-y\right)\left(1+{\left(y'\right)}^{2}\right)}}=\text{const}\tag{24}
$$

```python
u=sp.symbols('u',real=True); u1=sp.symbols('u_1',real=True)
x=sp.symbols('x',real=True); y=sp.Function('y')(x)
L=sp.sqrt((1+u1*u1)/(1-u))
eqy=sp.simplify(u1*sp.diff(L,u1)-L)
display(eqy.subs({u:y,u1:sp.diff(y,x)}))
```

Taking $y\left(0\right)=1$ and an initially vertical tangent, $\left|y\prime\left(0\right)\right|\to\infty$, suggests the substitution

$$
x\left(t\right)=t-\sin{t}, y=\cos{t}
$$

```python
theta=sp.symbols('theta',positive=True)
cycloid_beltrami=eqy.subs({u:sp.cos(theta),u1:-sp.sin(theta)/(1-sp.cos(theta))})
# For 0 < theta < pi both half-angle sine and cosine are positive.
display(sp.simplify(sp.trigsimp(cycloid_beltrami)))
display(sp.Eq(sp.Symbol('Beltrami_constant'),-1/sp.sqrt(2)))
```

The cycloid is also tautochronous in a uniform gravitational field: the oscillation period is independent of amplitude.

```python
y0=2.0; gravity=9.8; omega0=np.sqrt(gravity/(2*y0))
def source_cycloid_position(s):
    s=np.clip(np.asarray(s),0,1)
    return np.array([y0*(np.arcsin(s)-s*np.sqrt(1-s*s)),-y0*s*s])

def show_tautochrone(time):
    q=np.linspace(0,2*np.pi,600)
    fig,axes=plt.subplots(1,2,figsize=(10,4),layout='constrained')
    for ax in axes:
        ax.plot(y0*(q-np.sin(q))/2,-y0*np.sin(q/2)**2); ax.set(aspect='equal',xlabel='x',ylabel='y')
    s=abs(np.sin(omega0*time)); point=source_cycloid_position(s)
    if np.pi/(2*omega0)<=time<=3*np.pi/(2*omega0): point[0]=2*np.pi-point[0]
    axes[0].plot(*point,'ro'); axes[0].set_title('Motion along a cycloid')
    # Reproduce the source's illustrative interpolation for its ten points.
    # It is not the exact tautochrone dynamics; see the source review notes.
    s0=np.arange(1,11)/10; interpolated=s0+(1-s0)*abs(np.sin(omega0*time))
    points=source_cycloid_position(interpolated)
    axes[1].scatter(*points,c=s0,cmap='hsv'); axes[1].set_title('Source multi-point illustration')
    plt.show()
tautochrone_controls=animate_parameter(show_tautochrone,0,2*np.pi/omega0,0,0.03,'t')
```

## Example 3: double-pendulum configuration and Lagrangian

```python
def double_pendulum_configuration(a1=0.0,a2=0.0):
    p1=np.array([np.sin(a1),-np.cos(a1)]); p2=p1+np.array([np.sin(a2),-np.cos(a2)])
    fig,ax=plt.subplots(figsize=(5,5),layout='constrained')
    ax.plot([0,p1[0],p2[0]],[0,p1[1],p2[1]],'ko-')
    ax.plot([0,0],[0,-1],'--',color='gray'); ax.plot([p1[0],p1[0]],[p1[1]-0.5,p1[1]+0.5],'--',color='gray')
    ax.set(xlim=(-2,2),ylim=(-2.1,0.2),aspect='equal'); plt.show()
double_pendulum_configuration()
configuration_controls=widgets.interactive(double_pendulum_configuration,
    a1=widgets.FloatSlider(min=-np.pi/3,max=np.pi/3,step=0.02,value=0),
    a2=widgets.FloatSlider(min=-np.pi/3,max=np.pi/3,step=0.02,value=0))
display(configuration_controls)
```
